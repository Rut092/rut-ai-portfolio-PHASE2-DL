## Multplication Gate and Simple Derivatives

```python
import numpy as np

x = 3.0
y = -4.0
z = x * y

# i am assuming that error gradient is 2.0
dz = 2.0

dx = y * dz
dy = x * dz

print(f"gradient of x is {dx}")
print(f"gradient of y is {dy}")
```

## A Full-Neuron (Dot product + Bias + ReLU)

```python
X = np.array([1.0,2.0])
W = np.array([-0.5,0.3])
b = 0.5

dot_out = np.dot(X,W) + b
relu_out = max(0,dot_out)

print(f"dot product output is {dot_out}")
print(f"relu output is {relu_out}")

# Assuming the loss returns as 1.0
d_out = 1.0

dz = d_out*(1 if relu_out > 0 else 0)

db = dz * 1.0
dw = X * dz
dx = W * dz
print()
print(f"gradient of x is {dx}")
print(f"gradient of w is {dw}")
print(f"gradient of b is {db}")
```
