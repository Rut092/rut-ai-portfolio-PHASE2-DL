## Dropout

```python
class RobustClassifier(nn.Module):
    def __init__(self,dropout_rate):
        super(RobustClassifier,self).__init__()

        self.fc1 = nn.Linear(784,512)
        self.fc2 = nn.Linear(512,10)
        self.dropout = nn.Dropout(dropout_rate)
        self.relu = nn.ReLU()

    def forward(self,x):
        x = x.view(-1,784)
        x = self.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        return x


model = RobustClassifier(dropout_rate=0.5)
model.train() # <-- This internally ACTIVATES Dropout
model.eval()  # <-- This internally DISABLES Dropout (Scales weights perfectly)
```
